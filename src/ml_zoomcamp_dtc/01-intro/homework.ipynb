```python
X = df[df.origin.eq('Asia')][["vehicle_weight","model_year"]].head(7).values

XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
y= np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ (X.T @ y) 
w_ = np.linalg.lstsq(X, y, rcond=None)[0]
assert np.allclose(w,w_)
w , w_
```
